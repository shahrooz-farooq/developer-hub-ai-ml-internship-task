# Exploring and Visualizing Iris Dataset

This notebook demonstrates exploratory data analysis (EDA) on the Iris dataset.
We will load, explore, and visualize the data to understand its structure and patterns.

## Import Required Libraries

```python
# pandas is used for data handling (tables)
import pandas as pd

# seaborn is used for data visualization (graphs)
import seaborn as sns

# matplotlib is also used for plotting graphs
import matplotlib.pyplot as plt
```

## Load the Iris Dataset

Loading the built-in iris dataset from seaborn.
This dataset contains flowers data (3 species with measurements)

```python
# Loading the built-in iris dataset from seaborn
iris = sns.load_dataset('iris')
```

## Basic Data Checking (Understanding Data)

```python
# Print total number of rows and columns in dataset
print("Shape:", iris.shape)
```

```python
# Print column names (features of dataset)
print("\nColumns:")
print(iris.columns)
```

```python
# Show first 5 rows of dataset to understand data structure
print("\nFirst Rows:")
print(iris.head())
```

```python
# Show full information about dataset
# It includes data types and missing values info
print("\nInfo:")
print(iris.info())
```

```python
# Show statistical summary of dataset
# Includes mean, min, max, standard deviation etc.
print("\nStatistics:")
print(iris.describe())
```

## Data Visualization Section

### Scatter Plot

This graph shows relationship between sepal length and petal length.
Different colors (hue) show different flower species.

```python
sns.scatterplot(
    data=iris,
    x='sepal_length',
    y='petal_length',
    hue='species'
)

# Title for scatter plot
plt.title('Scatter Plot')

# Display the plot
plt.show()
```

### Histogram Plot

This shows distribution of all numerical columns.
Helps us understand how data is spread.

```python
# Histogram Plot
iris.hist(figsize=(10,8))

# Show histogram
plt.show()
```

### Box Plot

This graph is used to detect outliers in data.
It also shows median and data spread.

```python
# Box Plot
sns.boxplot(data=iris)

# Show box plot
plt.show()
```
